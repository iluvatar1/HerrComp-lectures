## Vibrations and waves

$$
\frac{\partial^2 u}{\partial t^2} = c^2 \nabla^2 u
$$

Unlike the heat equation, which describes irreversible spreading, the wave equation describes **reversible propagation at a finite speed** $c$. It is the typical **hyperbolic** equation: we need the initial displacement *and* the initial velocity, plus boundary conditions.

| System | Field $u$ | Wave speed $c$ |
|---|---|---|
| Vibrating string | transverse displacement | $\sqrt{T/\mu}$ ($T$ = tension, $\mu$ = mass per length) |
| Sound | pressure or density | $\approx 343$ m/s in air |
| Light | $\vec E$, $\vec B$ fields | $1/\sqrt{\epsilon_0\mu_0} \approx 3\times10^8$ m/s |
| Seismic P-waves | rock displacement | $\sim 6$ km/s in the crust |

### Derivation: a vibrating string

```{figure} https://upload.wikimedia.org/wikipedia/commons/1/15/StringParameters.svg
:width: 40%
:align: center
```

Take a small piece of string between $x$ and $x + dx$, with mass $\mu\,dx$, under tension $T$. If the string makes a small angle $\theta(x) \approx \partial u/\partial x$ with the horizontal, the net vertical force is $T\left[\sin\theta(x + dx) - \sin\theta(x)\right] \approx T\,\frac{\partial^2 u}{\partial x^2}\,dx$. Newton's second law gives

$$
\mu\,dx\,\frac{\partial^2 u}{\partial t^2} = T\,\frac{\partial^2 u}{\partial x^2}\,dx
\quad\Longrightarrow\quad
\frac{\partial^2 u}{\partial t^2} = c^2\,\frac{\partial^2 u}{\partial x^2}, \qquad c = \sqrt{T/\mu}.
$$

### D'Alembert's solution

On an infinite line every solution has the form

$$
u(x, t) = f(x - ct) + g(x + ct),
$$

a shape $f$ moving to the right and a shape $g$ moving to the left, both at speed $c$ and **without changing shape**. With zero initial velocity, $f = g = u(x, 0)/2$: the initial pulse splits into two halves. Information travels at exactly speed $c$, which will be the key to stability.

## The leap-frog scheme

With $x_i = i h$, $t_n = n\,\Delta t$ and centered second differences in both time and space:

$$
\frac{u_i^{n+1} - 2u_i^n + u_i^{n-1}}{\Delta t^2} = c^2\,\frac{u_{i+1}^n - 2u_i^n + u_{i-1}^n}{h^2}
\quad\Longrightarrow\quad
\boxed{u_i^{n+1} = 2u_i^n - u_i^{n-1} + r^2\left(u_{i+1}^n - 2u_i^n + u_{i-1}^n\right), \qquad r = \frac{c\,\Delta t}{h}.}
$$

The scheme is explicit and second order in both $\Delta t$ and $h$. The number $r$ is the **Courant number**.

```{figure} fig/wave-discreteSol.png
The leap-frog stencil uses two old time levels to compute the new one.
```

```{admonition} You have seen this before: molecular dynamics
:class: note
Write the semi-discrete equation as $\ddot u_i = \frac{c^2}{h^2}\left(u_{i+1} - 2u_i + u_{i-1}\right)$. This is Newton's law for a **chain of masses connected by springs**, each mass pulled by its two neighbours. The update above is exactly the Verlet/leap-frog integrator of the [molecular dynamics notebook](../08-MolecularDynamics/MolecularDynamics-MacroBodies.ipynb), $x^{n+1} = 2x^n - x^{n-1} + a^n\Delta t^2$. A string is the continuum limit of that chain, and leap-frog inherits its good energy behaviour.
```

### The first step

Leap-frog needs two time levels, but at $t = 0$ we only know $u_i^0$ and the velocity $v_i = \partial_t u(x_i, 0)$. The centered velocity $v_i \approx (u_i^1 - u_i^{-1})/(2\Delta t)$ defines a "ghost" level $u_i^{-1} = u_i^1 - 2\Delta t\, v_i$. Inserting it into the leap-frog formula for $n = 0$ gives a second-order first step:

$$
u_i^1 = u_i^0 + \Delta t\, v_i + \frac{r^2}{2}\left(u_{i+1}^0 - 2u_i^0 + u_{i-1}^0\right).
$$

### The solver

The solver below takes the initial displacement and velocity, the Courant number and the number of steps. Boundary conditions are given as a function `bc(u_new, u, u_prev, r)` that fixes the end values of the new time level; by default the end values are kept fixed. It returns the profiles every `nsave` steps.

````python
def leapfrog_step(u, u_prev, r):
    """One leap-frog step for the interior points; the end values are copied from u."""
    u_new = u.copy()
    u_new[1:-1] = 2*u[1:-1] - u_prev[1:-1] + r**2*(u[2:] - 2*u[1:-1] + u[:-2])
    return u_new


def solve_wave_1d(u0, v0, r, dt, nsteps, bc=None, nsave=1):
    """Leap-frog solution of u_tt = c^2 u_xx.

    u0, v0 : initial displacement and velocity (arrays)
    r      : Courant number c*dt/h;  dt: time step;  nsteps: number of steps
    bc     : function bc(u_new, u, u_prev, r) that sets the end values of u_new (default: fixed ends)
    Returns an array with the profiles at steps 0, nsave, 2*nsave, ...
    """
    u_prev = u0.copy()
    u = u0.copy()
    u[1:-1] = u0[1:-1] + dt*v0[1:-1] + 0.5*r**2*(u0[2:] - 2*u0[1:-1] + u0[:-2])   # first step
    frames = [u0.copy()]
    for n in range(1, nsteps + 1):          # here u holds the solution at step n
        if n % nsave == 0:
            frames.append(u.copy())
        if n == nsteps:
            break
        u_new = leapfrog_step(u, u_prev, r)
        if bc is not None:
            bc(u_new, u, u_prev, r)
        u_prev, u = u, u_new
    return np.array(frames)
````

### First example: a plucked string

A Gaussian bump starts at rest in the middle of a string with fixed ends ($L = 1$, $c = 1$, $N = 201$, $r = 0.9$). As d'Alembert predicts, it splits into two half-height pulses, which reflect at the fixed ends **upside down** and meet again at $t = 1$ inverted, and recover the initial shape at $t = 2$.

````python
N, c, r = 201, 1.0, 0.9
x = np.linspace(0, 1, N)
h = x[1] - x[0]
dt = r*h/c
u0 = np.exp(-(x - 0.5)**2/(2*0.05**2))
u0[0] = u0[-1] = 0.0
v0 = np.zeros(N)
nsteps = int(round(2.0/dt))
frames = solve_wave_1d(u0, v0, r, dt, nsteps)
times = dt*np.arange(len(frames))

fig, axes = plt.subplots(2, 3, figsize=(12, 5.5), sharex=True, sharey=True)
for ax, t_show in zip(axes.flat, [0, 0.2, 0.4, 0.6, 1.0, 2.0]):
    k = int(round(t_show/dt))
    ax.plot(x, frames[k], lw=1.5)
    ax.set_title(f"t = {times[k]:.2f}")
    ax.axhline(0, color="gray", lw=0.5)
for ax in axes[1]:
    ax.set_xlabel("$x$")
for ax in axes[:, 0]:
    ax.set_ylabel("$u$")
fig.suptitle(f"Plucked string with fixed ends (r = {r})")
plt.tight_layout()
plt.show()
````

## Stability: the CFL condition

### Von Neumann analysis

Insert a Fourier mode $u_i^n = G^n e^{\mathrm{i}kx_i}$, where $G = u^{n+1}/u^n$ is the amplification factor per step. With $s = \sin(kh/2)$, the leap-frog formula gives a quadratic equation for $G$:

$$
G^2 - 2\left(1 - 2r^2s^2\right)G + 1 = 0 .
$$

The product of its two roots is 1. If $r s \le 1$, the roots are complex conjugates with $|G| = 1$: the mode oscillates without growing or decaying, as a wave should. If $r s > 1$, the roots are real and one of them has $|G| > 1$: the mode grows exponentially. The worst mode is the saw-tooth ($s = 1$), so stability requires

$$
\boxed{r = \frac{c\,\Delta t}{h} \le 1} \qquad \text{(Courant–Friedrichs–Lewy condition).}
$$

In $d$ dimensions with equal spacings, the condition is $r \le 1/\sqrt d$.

**Physical meaning.** In one step, the numerical scheme can only move information by one cell, i.e. at speed $h/\Delta t$. The CFL condition says that this numerical speed must be at least the physical speed $c$: the numerical domain of dependence must contain the physical one.

Notice the difference with the heat equation: here $\Delta t \propto h$ (not $h^2$), which is much less restrictive.

### Seeing the instability

````python
N = 101
x = np.linspace(0, 1, N)
h = x[1] - x[0]
fig, ax = plt.subplots(figsize=(6, 4))
for r in [1.0, 1.02]:
    dt = r*h
    frames = solve_wave_1d(np.sin(np.pi*x), np.zeros(N), r, dt, 300)
    ax.semilogy(np.abs(frames).max(axis=1), label=f"r = {r}")
ax.set(xlabel="step", ylabel="max $|u|$", title="Leap-frog at and just above the CFL limit")
ax.legend()
plt.show()
````

## Validation

### Standing waves: an exact solution

With fixed ends, $u(x, 0) = \sin(\pi x)$ and zero velocity, the exact solution is the standing wave $u = \sin(\pi x)\cos(\pi c t)$, with period $T = 2L/c = 2$. We measure the error at a quarter period, $t = 0.5$, for several Courant numbers, and the order of accuracy with $r = 0.5$ fixed. (Measuring after exactly one full period would be misleading: there the error of the phase cancels to first order, and the error looks fourth order.)

````python
c = 1.0
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4))

N = 51
x = np.linspace(0, 1, N)
h = x[1] - x[0]
rs = [0.3, 0.5, 0.7, 0.9, 1.0]
errs_r = []
for r in rs:
    dt = r*h/c
    nsteps = int(round(0.5/dt))
    frames = solve_wave_1d(np.sin(np.pi*x), np.zeros(N), r, dt, nsteps)
    errs_r.append(np.max(np.abs(frames[-1] - np.sin(np.pi*x)*np.cos(np.pi*c*nsteps*dt))) + 1e-17)
ax1.semilogy(rs, errs_r, "o-")
ax1.set(xlabel="Courant number $r$", ylabel="max error at $t = 0.5$", title=f"N = {N}")

Ns = np.array([21, 41, 81, 161, 321])
errs_h = []
for N in Ns:
    x = np.linspace(0, 1, N)
    h = x[1] - x[0]
    dt = 0.5*h/c
    nsteps = int(round(0.5/dt))
    frames = solve_wave_1d(np.sin(np.pi*x), np.zeros(N), 0.5, dt, nsteps, nsave=nsteps)
    errs_h.append(np.max(np.abs(frames[-1] - np.sin(np.pi*x)*np.cos(np.pi*c*nsteps*dt))))
hs = 1/(Ns - 1)
slope = np.polyfit(np.log(hs), np.log(errs_h), 1)[0]
ax2.loglog(hs, errs_h, "o-", label=f"r = 0.5 (slope {slope:.2f})")
ax2.loglog(hs, errs_h[0]*(hs/hs[0])**2, "k--", lw=1, label=r"$\propto h^2$")
ax2.set_xticks(hs, [f"1/{N-1}" for N in Ns])
ax2.minorticks_off()
ax2.set(xlabel="$h$", ylabel="max error at $t = 0.5$", title="Order of accuracy")
ax2.legend()
plt.tight_layout()
plt.show()
````

The error drops to round-off level at $r = 1$. In 1D and only at $r = 1$, leap-frog reproduces d'Alembert's solution *exactly*: each value is copied one cell per step, which is precisely the exact motion. For $r < 1$ the scheme is second order.

### Numerical dispersion

The exact equation is non-dispersive: all wavelengths travel at speed $c$. The scheme is not. Inserting $u_i^n = e^{\mathrm{i}(kx_i - \omega t_n)}$ gives the numerical dispersion relation

$$
\omega_\text{num} = \frac{2}{\Delta t}\arcsin\!\left(r\,\sin\frac{kh}{2}\right),
$$

which tends to $\omega = ck$ for long waves ($kh \ll 1$) but makes short waves travel **slower** than $c$, unless $r = 1$.

````python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4))
kh = np.linspace(1e-6, np.pi, 300)
for r in [0.3, 0.7, 0.9, 1.0]:
    omega_dt = 2*np.arcsin(r*np.sin(kh/2))
    ax1.plot(kh/np.pi, (omega_dt/kh)/r, label=f"r = {r}")      # numerical phase speed / c
ax1.set(xlabel=r"$kh/\pi$", ylabel=r"$c_\mathrm{num}/c$", title="Numerical phase speed (leap-frog)")
ax1.legend()

# A very narrow pulse travelling to the right: u = f(x - ct), so v0 = -c f'(x)
N = 401
x = np.linspace(0, 1, N)
h = x[1] - x[0]
sigma, x0, t_end = 0.008, 0.2, 0.6
f = lambda s: np.exp(-(s - x0)**2/(2*sigma**2))
for r in [0.5, 1.0]:
    dt = r*h
    nsteps = int(round(t_end/dt))
    v0 = (x - x0)/sigma**2*f(x)                      # -c f'(x) with c = 1
    frames = solve_wave_1d(f(x), v0, r, dt, nsteps, nsave=nsteps)
    ax2.plot(x, frames[-1], label=f"r = {r}")
ax2.plot(x, f(x - t_end), "k--", lw=1, label="exact")
ax2.set(xlim=(0.65, 0.9), xlabel="$x$", ylabel="$u$", title=f"Narrow pulse after t = {t_end}")
ax2.legend()
plt.tight_layout()
plt.show()
````

With $r = 0.5$ the short-wave components of the narrow pulse lag behind and form a trail of ripples behind the pulse. Resolving the pulse with more points per width ($\sigma/h \gtrsim 10$) removes the problem.

### Energy

The energy of the string, $E = \int \left[\tfrac12 u_t^2 + \tfrac12 c^2 u_x^2\right]dx$, is conserved by the exact equation. Leap-frog does not dissipate energy (unlike the heat equation, where decay is the physics): a discrete version of $E$ oscillates slightly but does not drift, just like the energy of a Verlet integration in molecular dynamics. You will check it in an exercise.

## Boundary conditions and reflections

| Boundary | Physical example | Condition | Reflection |
|---|---|---|---|
| Fixed (Dirichlet) | string clamped to a wall | $u = 0$ | inverted |
| Free (Neumann) | string end on a frictionless ring, open pipe | $u_x = 0$ | upright |
| Periodic | ring, or a crystal | $u(0) = u(L)$ | none: the wave re-enters on the other side |
| Absorbing | infinite string, open space | outgoing wave only | (ideally) none |

For the boundary demos we send a single pulse to the right. For a right-moving wave $u = f(x - ct)$, the initial velocity is $v_0 = -c f'(x)$. With fixed ends (the default), the reflected pulse comes back inverted. At $t = 0.7$ the pulse sits on the wall: think of the reflection as an "image" pulse coming from outside, inverted for a fixed end (so the two cancel and $u = 0$ at the wall) and upright for a free end (so they add up to height 2).

````python
N, c, r = 201, 1.0, 0.9
x = np.linspace(0, 1, N)
h = x[1] - x[0]
dt = r*h/c
x0, sigma = 0.3, 0.04
u0_right = np.exp(-(x - x0)**2/(2*sigma**2))
v0_right = c*(x - x0)/sigma**2*u0_right              # -c f'(x)


def plot_reflection(frames, title):
    """Snapshots of a right-moving pulse before and after it reaches the right end."""
    fig, ax = plt.subplots(figsize=(7, 3.5))
    for t_show in [0.0, 0.4, 0.7, 1.0]:
        k = int(round(t_show/dt))
        ax.plot(x, frames[k], label=f"t = {t_show}")
    ax.set(xlabel="$x$", ylabel="$u$", title=title, ylim=(-2.1, 2.1))
    ax.legend(fontsize=8, loc="lower left")
    plt.tight_layout()
    plt.show()


frames = solve_wave_1d(u0_right, v0_right, r, dt, int(round(1.0/dt)))
plot_reflection(frames, "Fixed end: inverted reflection")
````

## Two dimensions

For a membrane, $u_{tt} = c^2(u_{xx} + u_{yy})$. With $h_x = h_y = h$, leap-frog uses the five-point stencil again:

$$
u_{i,j}^{n+1} = 2u_{i,j}^n - u_{i,j}^{n-1} + r^2\left(u_{i+1,j}^n + u_{i-1,j}^n + u_{i,j+1}^n + u_{i,j-1}^n - 4u_{i,j}^n\right), \qquad r \le \frac{1}{\sqrt2}.
$$

```{figure} fig/wave-2d-points.png
:align: center
```

````python
def laplacian5(u):
    """Five-point sum u[i+1,j] + u[i-1,j] + u[i,j+1] + u[i,j-1] - 4u[i,j] on the interior."""
    return u[2:, 1:-1] + u[:-2, 1:-1] + u[1:-1, 2:] + u[1:-1, :-2] - 4*u[1:-1, 1:-1]


def solve_wave_2d(u0, r, nsteps, nsave=1):
    """Leap-frog for the 2D wave equation: zero initial velocity, u = 0 on the edges."""
    u_prev = u0.copy()
    u = u0.copy()
    u[1:-1, 1:-1] = u0[1:-1, 1:-1] + 0.5*r**2*laplacian5(u0)       # first step (v0 = 0)
    frames = [u0.copy()]
    for n in range(1, nsteps + 1):
        if n % nsave == 0:
            frames.append(u.copy())
        if n == nsteps:
            break
        u_new = np.zeros_like(u)
        u_new[1:-1, 1:-1] = 2*u[1:-1, 1:-1] - u_prev[1:-1, 1:-1] + r**2*laplacian5(u)
        u_prev, u = u, u_new
    return np.array(frames)


N, c, r = 101, 1.0, 0.5
x = np.linspace(0, 1, N)
h = x[1] - x[0]
dt = r*h/c
XX, YY = np.meshgrid(x, x, indexing="ij")
u0 = np.exp(-((XX - 0.4)**2 + (YY - 0.5)**2)/(2*0.04**2))
u0[0, :] = u0[-1, :] = u0[:, 0] = u0[:, -1] = 0.0
frames = solve_wave_2d(u0, r, 320, nsave=8)
animate_2d(frames, dt*8*np.arange(len(frames)))
````

Notice two differences with 1D: the circular wave **decreases in amplitude** as it spreads (the energy is spread over a growing circle, so $u \sim 1/\sqrt{\rho}$), and it leaves a **trailing wake** behind the front. In 2D the sharp-front property of d'Alembert's solution is lost (it holds again in 3D: Huygens' principle).

### Normal modes of a square drum

A membrane with fixed edges on $[0, L]^2$ has the normal modes

$$
u_{mn}(x, y, t) = \sin\frac{m\pi x}{L}\,\sin\frac{n\pi y}{L}\,\cos(\omega_{mn}t), \qquad \omega_{mn} = \frac{\pi c}{L}\sqrt{m^2 + n^2}.
$$

Unlike a string, the frequencies are *not* multiples of a fundamental (which is why a drum sounds less "musical" than a guitar), and different modes can have the same frequency: $(1,2)$ and $(2,1)$ are **degenerate**, like states in quantum mechanics. See [the modes of a drum](https://www.youtube.com/watch?v=7nG4Y7FU-dQ).

## Summary

| Topic | Key result |
|---|---|
| Scheme | leap-frog: $u^{n+1} = 2u^n - u^{n-1} + r^2\,\delta_x^2 u^n$, second order; special first step from $v_0$ |
| Stability | CFL: $r = c\Delta t/h \le 1$ (1D), $\le 1/\sqrt d$ in $d$ dimensions; $\Delta t \propto h$ |
| Accuracy | $O(h^2)$; exact in 1D at $r = 1$; short waves travel too slowly (numerical dispersion) |
| Energy | not dissipated: bounded oscillation, as in Verlet/MD |
| Boundaries | fixed: inverted reflection; free: upright; periodic: re-entry; Mur: absorbing |
| 2D | amplitude decays as the wave spreads; drum modes $\omega_{mn} \propto \sqrt{m^2 + n^2}$, with degeneracies |

Comparison of the three model equations of this chapter:

| | Laplace (elliptic) | Heat (parabolic) | Wave (hyperbolic) |
|---|---|---|---|
| Needs | boundary values | initial $u$ + boundaries | initial $u$ and $u_t$ + boundaries |
| Typical method | relaxation, sparse solve | FTCS, implicit θ-method | leap-frog |
| Explicit stability | — | $\Delta t \le h^2/(2\alpha)$ | $\Delta t \le h/c$ |
| Behaviour | smooth, no time | smoothing, irreversible | propagation, reversible |
